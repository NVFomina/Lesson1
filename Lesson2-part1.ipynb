```python
triples = [(1, -12, 20), (1, 17, 72), (1, -7, -44), (1, 9, 8), (1, -2, -63)]
for a, b, c in triples:
    discr = b**2 - 4*a*c
    if discr > 0:
        print((-b + discr**.5)/2*a, (-b - discr**.5)/2*a)
    else:
        if discr == 0:
            print(-b/2*a)
        else:
            print('no solution')
```

```python
lst = [1, 2, 3, 4, 5]

def change(lst):
    first_elem = lst.pop(0)
    last_elem = lst.pop(len(lst)-1)
    lst.insert(0, last_elem)
    lst.append(first_elem)
    print(lst)

change(lst)
```
